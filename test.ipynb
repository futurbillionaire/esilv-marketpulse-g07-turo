```python
def display_market_summary(asset, prices, show_currency=True):
    """Affiche un résumé de marché pour n'importe quel actif (action ou indice)."""
    ticker = asset["ticker"]
    rows = [r for r in prices if r.get("ticker", ticker) == ticker]
    if rows and "date" in rows[0]:
        rows = sorted(rows, key=lambda r: r["date"])

    if not rows:
        print(f"{ticker} - {asset['name']}")
        print("Aucune donnée pour cet actif.")
        return

    first_close = rows[0]["close"]
    last_close = rows[-1]["close"]

    currency = f" {asset['currency']}" if show_currency and asset.get("currency") else ""

    print(f"{ticker} - {asset['name']}")
    print(f"Observations : {len(rows)}")
    print(f"First close  : {first_close:.2f}{currency}")
    print(f"Last close   : {last_close:.2f}{currency}")

from datetime import date, timedelta

def make_prices(ticker, start, end, n=21):
    step = (end - start) / (n - 1)
    d0 = date(2026, 9, 1)
    return [
        {"ticker": ticker, "date": d0 + timedelta(days=i), "close": start + i * step}
        for i in range(n)
    ]

instrument = {"ticker": "AAPL", "name": "Apple Inc.", "currency": "USD"}
benchmark = {"ticker": "SP500", "name": "S&P 500"}

instrument_prices = make_prices("AAPL", 250.00, 266.20)
benchmark_prices = make_prices("SP500", 6600.00, 6742.00)

# ---------- Même fonction, données différentes ----------
display_market_summary(instrument, instrument_prices)
print()
display_market_summary(benchmark, benchmark_prices, show_currency=False)
```
